```python
import tensorflow as tf
from tensorflow import keras
import matplotlib.pyplot as plt

%matplotlip inline
import numpy as np
```

```python
(X_train, y_train), (X_test, y_test) = keras.datasets.mnist.load_data()
```

```python
print(len(X_train), len(X_test), len(y_train), len(y_test))
```

```python
# flatten the X_train values from a 28 x 28 shape to a vector
X_train_flat = X_train.reshape(-1, 28 * 28) # or X_train_flat = X_train.reshape(len(X_train), 28 * 28)

#... and flatten the test set
X_test_flat = X_test.reshape(-1, 28 * 28) # or X_test_flat = X_test.reshape(len(X_test), 28 * 28)

print(X_train_flat.shape, X_test_flat.shape)
```

```python
# scale the samples
X_train = X_train/255
X_test = X_test/255
```

```python
# create a neural network with 784 input neurons and an output of 10 neurons
# 'Dense' means each input neuron is connected to a neuron in the output layer
model = keras.Sequential([
        keras.layers.Dense(10, activation='sigmoid', input_shape=(784,))
])

# compile the neural network model
model.compile(optimizer='adam',
              loss='sparse_categorical_crossentropy',
              metrics=['accuracy'])

model.fit(X_train_flat, y_train, epochs=5)
```

# Module: tf.keras.losses

```python
model.evaluate(X_test_flat, y_test)
```
