```python
def get_bracket_nesting_levels(s: str):
    """
    Identifies the nesting levels of opened and closed brackets in a string.

    Args:
        s: The input string.

    Returns:
        A list of tuples, where each tuple contains:
        (bracket_character, index, level, type_of_bracket)
        - bracket_character: The bracket itself (e.g., '(', ']', '{').
        - index: The 0-based index of the bracket in the string.
        - level: The nesting level (0 for outermost, increasing with depth).
        - type_of_bracket: 'opened' or 'closed'.
    """
    bracket_map = {
        '(': ')',
        '[': ']',
        '{': '}',
        ')': '(',
        ']': '[',
        '}': '{'
    }
    opened_brackets = set(['(', '[', '{'])
    closed_brackets = set([')', ']', '}'])

    level_stack = []  # Stores the current nesting level
    result = []
    current_level = 0

    for i, char in enumerate(s):
        if char in opened_brackets:
            current_level += 1
            result.append((char, i, current_level, 'opened'))
            level_stack.append(current_level)
        elif char in closed_brackets:
            if level_stack:
                # Get the level of the corresponding opened bracket
                # and assign it to the closed bracket
                level_for_closed = level_stack.pop()
                result.append((char, i, level_for_closed, 'closed'))
                current_level = level_for_closed -1
            else:
                # Handle cases where a closed bracket appears without an open one
                result.append((char, i, 0, 'closed')) # Assign level 0 or handle as error
    return result

# Example Usage:
if __name__ == "__main__":
    test_strings = [
        "abc(def[ghi]jkl)mno",
        "{a[b(c)d]e}f",
        "[[[]]]",
        "()",
        "abc",
        "]",
        "([)]", # Mismatched brackets
        "{(())}[()]"
    ]

    for s in test_strings:
        print(f"String: '{s}'")
        levels = get_bracket_nesting_levels(s)
        for bracket, index, level, bracket_type in levels:
            print(f"  Character: '{bracket}', Index: {index}, Level: {level}, Type: {bracket_type}")
        print("-" * 30)

    # More complex example
    s_complex = "func(arg1, [list_item1, {key: 'value', nested_list: [1,2]}], arg2)"
    print(f"String: '{s_complex}'")
    levels_complex = get_bracket_nesting_levels(s_complex)
    for bracket, index, level, bracket_type in levels_complex:
        print(f"  Character: '{bracket}', Index: {index}, Level: {level}, Type: {bracket_type}")
    print("-" * 30)
```
