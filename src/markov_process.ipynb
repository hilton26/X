```python
import numpy as np

# Define states
states = ['Sunny', 'Rainy']
state_to_index = {state: i for i, state in enumerate(states)}
index_to_state = {i: state for i, state in enumerate(states)}

# Define transition matrix
# Rows: Current state, Columns: Next state
# P[i, j] = probability of transitioning from state i to state j
transition_matrix = np.array([
    [0.8, 0.2],  # From Sunny: 80% Sunny, 20% Rainy
    [0.5, 0.5]   # From Rainy: 50% Sunny, 50% Rainy
])

# Simulate the Markov chain
def simulate_markov_chain(initial_state, num_steps):
    current_state_index = state_to_index[initial_state]
    path = [initial_state]

    for _ in range(num_steps):
        # Get probabilities for the current state
        probabilities = transition_matrix[current_state_index]
        # Choose the next state based on probabilities
        next_state_index = np.random.choice(len(states), p=probabilities)
        current_state_index = next_state_index
        path.append(index_to_state[current_state_index])
    return path

# Run a simulation
initial_state = 'Sunny'
num_steps = 10
simulated_path = simulate_markov_chain(initial_state, num_steps)
print(f"Simulated path: {simulated_path}")
```
