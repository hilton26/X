#### From a list of number get those which add up to a given sum
https://youtu.be/9T-K7rkOCAo

```python
nums = [6, 4, 7, 9, 3, 12, 3]
t = 9
```

```python
for i in range(len(nums)):
    for j in range(len(nums)):
        if nums[i] != nums[j]:
            if nums[i] + nums[j] == t:
                print([nums[i], nums[j]])
```
