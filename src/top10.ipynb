```python
import pandas as pd
```

```python
data = pd.read_excel(r'C:\Users\hilton.netta\Downloads\poster.xlsx', sheet_name = 'tab', usecols = 'A:I').dropna(subset = 'Investment Type')
df = pd.DataFrame(data, columns=['Investment Type', 'i Issue Name', 'Primary Asset ID', 'Closing Exposure PA'])
df = df.rename(columns={'Investment Type': 'Type', 'Primary Asset ID': 'Asset ID', 'i Issue Name': 'Description', 'Closing Exposure PA': 'Exposure (ZAR)'})
df_top10 = df.loc[(~df['Type'].isin(['CASH', 'SYTH'])) & (~df['Description'].str.contains('REPO'))\
].nlargest(10, 'Exposure (ZAR)').reset_index(drop = True)
df_top10
```
