```python
# create a time difference function
import datetime, time

def timediff(start, end, decimals = 1):
    if   int((end - start) / 3600) > 0: # non-zero hours
        return str(int(  (end - start) / 3600))           + 'hr '  + \
               str(int(  (end - start) /   60))           + 'min ' + \
               str(round((end - start) %   60, decimals)) + 'sec'
    elif int((end - start) /   60) > 0: # non-zero hours and minutes
        return str(int(  (end - start) /   60))           + 'min ' + \
               str(round((end - start) %   60, decimals)) + 'sec'
    else:
        return str(round((end - start) %   60, decimals)) + 'sec'
```

```python
# connect to Outlook
start_time = time.time()
print('Connecting to Outlook ...')

outlook = win32.Dispatch("Outlook.Application").GetNamespace("MAPI") # connect to the Microsft Outlook messaging API,
                                                                     # MAPI = Mapping Application Programming Interface

print(f'Connecting to Outlook completed: {timediff(start_time, time.time())}', '\n')
```

```python
# connect to Outlook folder
start_time = time.time()
print('Connecting to Outlook folder ...')

inbox = outlook.GetDefaultFolder(6) # 6 = Inbox folder (without mails from the subfolder)
                                    # https://learn.microsoft.com/en-us/office/vba/api/outlook.oldefaultfolders
                                    # deleted items = 3, outbox = 4, sent mail = 5, inbox = 6, drafts = 6, folder junk = 23

print(f'Connecting to Outlook folder completed: {timediff(start_time, time.time())}', '\n')
```

```python
# get messages
start_time = time.time()
print('Getting Outlook folder messages ...')

messages = inbox.Items

for message in messages:
    subject     = message.Subject
    body        = message.body
    attachments = message.Attachments
    
    # create a separate folder for each message
    #target_folder = output_dir / str(subject)
    target_folder = output_dir / str('subject')
    target_folder.mkdir(parents = True, exist_ok = True)
    
    # write email body to text file
    Path(target_folder / "EMAIL_BODY.txt").write_text(str(body))
    
    # save attachments
    for attachment in attachments:
        attachment.SaveAsFile(target_folder) / str(attachment)
    

print(f'Getting Outlook folder messages completed: {timediff(start_time, time.time())}', '\n')
```
