```python
import pandas as pd

# Create a sample DataFrame
data = {'Col1': [1.2, 3.4, 5.5, float('nan')],
        'Col2': [10.1, 20.2, 30.3, 40.4]}
df = pd.DataFrame(data)

print("Original DataFrame:")
print(df)
print("\nOriginal Data Types:")
print(df.dtypes)

# Convert 'Col1' from float to string
df['Col1'] = df['Col1'].astype(str)

print("\nDataFrame after conversion:")
print(df)
print("\nNew Data Types:")
print(df.dtypes)
```

```python
import pandas as pd
import numpy as np

s = pd.Series([1, 2, np.nan, 4])

# Case 1: NaN not in the values list
result_1 = s.isin([1, 2])
print("Result 1 (NaN not in values list):\n", result_1)

# Case 2: NaN in the values list
result_2 = s.isin([1, 2, np.nan])
print("\nResult 2 (NaN in values list):\n", result_2)
```
