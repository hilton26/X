```python
# Moonfurr automator
# Start an external program: https://stackoverflow.com/questions/37238645/how-to-open-external-programs-in-python

#!pip install pyautogui
import pyautogui as pya
import subprocess
import time

def moonfurr(exe=r"M:\Generator\Generator.exe", generator_run_time_in_sec=15):
    start_time = time.time()

    # Start the Moonfire application
    subprocess.Popen([exe, "/start", "/max"])

    # Select Reports tab
    time.sleep(10)  # give the app time to complete its opening sequence
    pya.click(1000, 500)  # bring the app window into focus
    pya.hotkey("ctrl", "n")  # open the report menu

    time.sleep(3)  # give the app time to complete its opening sequence
    pya.click(1000, 500)  # bring the app window into focus
    pya.typewrite(
        ["\t", "\t", "\t", "\t", "\t", "\t", "\t", "\t"]
    )  # go to the "Look Through Units" bullet
    pya.typewrite(["space"])  # select the "Look Through Units" bullet
    pya.typewrite(
        ["\t", "\t", "\t", "\t", "\t", "\t"]
    )  # go to the first item under Available Reports
    pya.typewrite(["space"])  # select "ASISA Manager Format Report"
    pya.doubleClick(800, 205)  # go to "ECICBALC" and double click it
    pya.typewrite(["\t", "\t"])  # go to the "Next" tab
    pya.typewrite(["\n"])  # press "Enter" on the "Next" button

    # Reporting Period tab
    # time.sleep(3)
    pya.click(1000, 500)
    pya.typewrite(["\t", "\t", "\t", "\t", "\t", "\t", "\t"])
    pya.typewrite(["down", "down", "down"])
    pya.typewrite(["\t", "\t", "\t", "\t"])  # tab to the "Okay" button
    pya.typewrite(["\n"])  # press "Enter" on the "Okay" button

    pya.click(1000, 500)
    pya.typewrite(["\t", "\t", "\t", "\t", "\t", "\t"])
    pya.typewrite(["\n"])  # press "Enter" on the "Okay" button

    # Portfolios to Include tab
    time.sleep(1)
    pya.click(1000, 500)
    pya.typewrite(["\t", "\t"])  # tab to "Portfolios"
    pya.typewrite(["down"])  # go one down to "Fund Check"
    pya.hotkey("shift", "right")  # select "Fund Check"
    pya.typewrite(
        [
            "down",
            "down",
            "down",
            "down",
            "down",
            "down",
            "down",
            "down",
            "down",
            "down",
            "down",
            "down",
            "down",
            "down",
            "down",
            "down",
            "down",
            "down",
            "down",
            "down",
            "down",
            "down",
            "down",
            "down",
            "down",
            "down",
            "down",
            "down",
            "down",
            "down",
            "down",
            "down",
            "down",
            "down",
            "down",
            "down",
            "down",
            "down",
            "down",
            "down",
            "down",
        ]
    )  # go down to "Prescient Other Funds"
    time.sleep(1)
    pya.typewrite(
        ["\t", "\t", "\t", "\t", "\t", "\t", "\t"]
    )  # tab until "ECICBALC" appears under PortfolioCode
    pya.typewrite(["space"])  # select "ECICBALC" under PortfolioCode
    pya.doubleClick(800, 210)  # go to "ECICBALC" and double-click it

    pya.click(1000, 500)  # ensure focus is on the app window
    pya.typewrite(
        ["\t", "\t", "\t", "\t", "\t", "\t"]
    )  # tab until "Okay" button is in focus
    pya.typewrite(["\n"])  # press "Enter" on the "Okay" button

    # Groups and Categories tab
    time.sleep(1)
    pya.click(1000, 500)
    pya.typewrite(["\t", "\t", "\t", "\t"])  # tab to the "Okay" button
    pya.typewrite(["\n"])  # press "Enter" on the "Okay" button

    # Selected Criteria tab
    time.sleep(1)
    pya.click(1000, 500)
    pya.typewrite(["\t", "\t", "\t"])  # tab to the "Generate" button
    pya.typewrite(["\n"])  # press "Enter" on the "Generate" button

    # Generator Progress Report tab
    time.sleep(
        generator_run_time_in_sec
    )  # allowing 15 seconds for the report generation to complete
    pya.click(1000, 500)
    pya.typewrite(["\t"])  # tab to the "Okay" button
    pya.typewrite(["\n"])  # press "Enter" on the "Okay" button

    # Close the app
    pya.click(1000, 500)
    pya.hotkey("ctrl", "q")  # exit the application

    print(
        timediff(start_time, time.time()),
        f" to complete the run including {generator_run_time_in_sec}sec to generate the report",
    )
```
