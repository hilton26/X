```python
# get MSCI and Jxxx data from P:\Investment Operations\GRC\Compliance\MCaps\Ix.xlsm
start_time = time.time()

msci_zips    = r'P:\Investment Operations\GRC\Compliance\Reporting Requirements\PIM PPSBAL\Zips'
m = os.listdir(msci_zips)                                                     # list of files in the Zips directory
# k = max([x[-8:-4] + x[-12:-10] + x[-10:-8] for x in m if x.endswith('.zip')]) # sub-list of files ending in ".zip"
k = max([x[-12:-4] for x in m if x.endswith('.zip')]) # sub-list of files ending in ".zip"
# zip_file     = f'PPS_MSCI_indexes_{k[4:6] + k[-2:] + k[0:4]}.zip'                                                                  
zip_file     = f'PPS_MSCI_Indices_{k}.zip' # latest zip file in the folder
zipped_file  = os.path.join(msci_zips, zip_file)

# TEST
print('',zip_file,'\n',zipped_file)

# find the .cn1 file in the zipped archive
with zipfile.ZipFile(zipped_file, mode="r") as archive:
     for filename in archive.namelist():
         if filename.endswith('.cn1'):
             cn1 = filename
             break

print('', f'{zip_file} is the latest zipped archive')
print('', f'{cn1} is the file in the archive to be read','\n')

# extract bytes data from the zipped .cn1 file
try:
    x = []
    with zipfile.ZipFile(zipped_file, mode="r") as archive:
        with archive.open(cn1, mode="r") as cn1:
            for line in cn1:
                # print(line.split(b','))
                x.append(line.split(b','))
except zipfile.BadZipFile as error:
    print(error)

# dataframe the retrieved bytes
df = pd.DataFrame(x)

# convert bytes to strings
def de_byte(byt):
    """ function to decode bytes to str """
    return byt.decode('utf-8')

df = df.map(de_byte)

# make the first row the header row
# https://stackoverflow.com/questions/31328861/replacing-header-with-top-row
new_header = df.iloc[0] #grab the first row for the header
df         = df[1:] #take the data less the header row
df.columns = new_header #set the header row as the df header

# drop the source sub-header row 
df.drop(1, inplace = True)

# reset the index
df.reset_index(level = None, drop = True, inplace = True)

# dataframe MSCI_all
df1 = df

# convert some columns from type string to type float
headings = ['market cap 1', 'market cap 2', 'weight 1', 'price']
df[headings] = df[headings].astype(float)

# remove "\r/n" from currency column entries
df = df.rename(columns = {'price currency\r\n': 'price currency'})
def remove_rn(txt):
    return txt.rstrip('\r\n')

df['price currency'] = df['price currency'].map(remove_rn)

idx_date = datetime.strptime(df["end date"].iloc[0],"%Y%m%d")

# use specific headingsfor the dataframe of indices
msci_heads = ['end date', 'series long name', 'Statpro ISIN', 'security name', 'price currency', 'market cap 1', 'weight 1']
df = df[msci_heads]

# get the unique sets of indices
series = df['series long name'].unique()
dict  = {'msci_wo': 'WORLD SELECTION', 'msci_em': 'EM SELECTION', 'msci_ac': 'AC WORLD', 'msci_gl': 'THE WORLD INDEX'}

print('', f'{len(series)} indices for {idx_date.strftime("%a %d %b %Y")} over {len(df)} rows: {(", ").join(series)}')
print('\n', f'{timediff(start_time, time.time())} to get zipped data')
```

```python
# write the MSCI dataframe df1 to the MSCI_ALL sheet of the Ix.xlsm file using win32.com

# # https://www.reddit.com/r/learnprogramming/comments/11ny2oe/writing_dataframes_to_excel_workbook_with/
# Use the win32com package to write the DataFrame to Excel using the Excel COM interface. This can be more reliable than using Pandas
# to write to the Excel file. Here's an example:
start_time = time.time()
print('','Writing to Ix.xlsm')

# get MSCi securities with unique ISINs
# https://stackoverflow.com/questions/43184491/df-unique-on-whole-dataframe-based-on-a-column
df1 = df1.drop_duplicates(subset = ['isin'])

# Open the Excel application
# !pip install pywin32
import win32com.client as win32 
excel = win32.gencache.EnsureDispatch('Excel.Application')
excel.Visible = False

# open the workbook and select the worksheet
wb = excel.Workbooks.Open(r'P:\Investment Operations\GRC\Compliance\MCaps\Ix.xlsm')
ws = wb.Worksheets('MSCI_ALL')

# clear the contents of the MSCI_ALL sheet
ws.Cells.ClearContents()

# write the MSCI headings to the first row of the MSCI_ALL sheet
ws.Range(ws.Cells(1,1), ws.Cells(1, len(df1.columns))).Value = list(df1)

# write the indices to the MSCI_ALL sheet
ws.Range(ws.Cells(2,1), ws.Cells(len(df1.index) + 1, len(df1.columns))).Value = df1.values.tolist()

# save and close the workbook and then quit the Excel application
wb.Save()
wb.Close()
# excel.Visible = True
# excel.Quit()

print('\n', f'{timediff(start_time, time.time())} writing to Ix.xlsm over {len(df1.index)} rows and {len(df1.columns)} columns')
```

```python
# write the World and EM indices to !IndexPrep.xlsm
start_time = time.time()
print('Updating !IndexPrep with MSCIWo and MSCIEM indices')

# isolate the WO and EM indices
# https://stackoverflow.com/questions/17957890/pandas-select-from-dataframe-using-startswith
df_WO = df[df['series long name'].str.startswith(r'MSCI/WO')]
df_EM = df[df['series long name'].str.startswith(r'MSCI/EM')]

pthIndexPrep = r'P:\Investment Operations\GRC\Compliance\Reporting Requirements\!IndexPrep.xlsm'

# open the workbook and select the worksheet
wb   = excel.Workbooks.Open(pthIndexPrep)
wsWO = wb.Worksheets('MSCIWo')
wsEM = wb.Worksheets('MSCIEM')

# clear the contents of the MSCIWo and MSCIEM sheets
wsWO.Range(wsWO.Cells(1,1), wsWO.Cells(1, len(df_WO.columns))).ClearContents()
wsEM.Range(wsEM.Cells(1,1), wsEM.Cells(1, len(df_WO.columns))).ClearContents()

# write the MSCI headings to the first row of the MSCIWo and MSCIEM sheets
wsWO.Range(wsWO.Cells(1,1), wsWO.Cells(1,len(df_WO.columns))).Value = list(df_WO)
wsEM.Range(wsEM.Cells(1,1), wsEM.Cells(1,len(df_EM.columns))).Value = list(df_EM)

# write the indices to the MSCIWo and MSCIEM sheets
wsWO.Range(wsWO.Cells(2,1), wsWO.Cells(len(df_WO.index) + 1, len(df_WO.columns))).Value = df_WO.values.tolist()
wsEM.Range(wsEM.Cells(2,1), wsEM.Cells(len(df_EM.index) + 1, len(df_EM.columns))).Value = df_EM.values.tolist()

# save and close the workbook and then quit the Excel application
wb.Save()
wb.Close()
excel.Visible = True
excel.Quit()

print('\n', f'{timediff(start_time, time.time())} updating !IndexPrep with MSCIWo and MSCIEM indices')
print(f'Roundtrip time to complete accessing zipped file and saving it: {timediff(start_time_0, time.time())}', '\n')
print(pthIndexPrep)
```
