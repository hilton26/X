```python
# Import Libraries

%run utilities.ipynb
import docx
import re
import os
from tqdm import tqdm
#from Word_Hyperlinks import get_hyperlinks

# import xlwt # xlwt is a library for writing data and formatting information to older Excel files (ie: .xls) https://xlwt.readthedocs.io/en/latest/
```

```python
# Constants

pth       = r'P:\Investment Operations\GRC\Compliance\Client Mandates'
file_save = f'PIM_Mandates_with_infrastructure {datetime.now().strftime("%a%b%y")}.xlsx'
keywords  = ['infra', 'infrastructure', 'Infra', 'Infrastructure']
```

```python
# Functions

def get_text(filename):
#get all the text in a docx given its file name
    doc = docx.Document(os.path.join(pth, filename))
    fullText = []
    for para in doc.paragraphs:
        fullText.append(para.text)
    return fullText

def square_bracketless(filename):
#get the text from a docx with internal square brackets removed
    g = get_text(filename)
    new_text = []
    for m in g:
        if '[' in m:
            tru = re.search('\[(.*?)\]',str(m)).group(0)
            new_text.append(m.replace(tru,''))
        else:
            new_text.append(m)
    return new_text

def get_clauses(keywds,fund):
#get the clauses from a square-bracketless file having matching keywords
    mumble = [fund.replace(" Rules.docx",""),]
    for i in get_text(fund):
            for j in keywds:
                if j in i:
                    if i in mumble:
                        pass
                    else:
                        mumble.append(i)
    return mumble
```

```python
# list all the rules docx files to be interrogated
start_time = time.time()
# iterate over the files in the directory
mandates = []
for i in os.listdir(pth):
    if os.path.isfile(os.path.join(pth, i)) and i.endswith(".docx") and not i.startswith("~") and not i.startswith("!"):
        # get_text(pth + i)
        mandates.append(i)
        
# print(len(list_of_mandates), list_of_mandates)
print(f'{timediff(start_time, time.time())}: to list {len(mandates)} mandate docx files')
```

```python
# Set up an xlsx workbook to save results - import openpyxl # http://automatetheboringstuff.com/chapter12/
start_time = time.time()
from openpyxl import Workbook
wb = Workbook()
ws = wb.active
print(f'{timediff(start_time, time.time())}: to open an openpyxl workbook')
```

```python
# Run through erasch mandate and store the results in a sheet
start_time = time.time()
for i, mandate in enumerate(tqdm(mandates)): # https://stackoverflow.com/questions/48437189/python-enumerate-tqdm-progress-bar-when-reading-a-file
    ws['A' + str(i + 1)].value = str(get_clauses(keywords, mandate))
    #ws['B' + str(i + 1)].value = str(get_hyperlinks('Segregated Clients', os.path.join(pth, mandate)))

wb.save(                      os.path.join(pth, file_save))
os.system(f'start EXCEL.EXE "{os.path.join(pth, file_save)}"') # https://stackoverflow.com/questions/35940748/use-python-to-launch-excel-file
print(f'{timediff(start_time, time.time())}: to run through {len(mandates)} mandate docx files. Results stored here: {os.path.join(pth, file_save)}')
```
