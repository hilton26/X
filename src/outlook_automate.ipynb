```python
import pandas as pd
import win32com.client as w32
from datetime import datetime, timedelta
```

```python
# connect to Outlook
outlook = w32.Dispatch("Outlook.Application").GetNamespace("MAPI")
inbox   = outlook.GetDefaultFolder(6) # 6 refers to the Inbox
```

```python
# calculate the date for a week ago
last_week     = datetime.now() - timedelta(days=7)
filter_string = '[ReceivedTime] >= "' + last_week.strftime("%m/%d/%Y") + '"'
filter_string
```

```python
# apply filter to get emails from last week
filtered_emails = inbox.Items.Restrict(filter_string)
```

```python
# for email in filtered_emails:
#     print(email)
```

```python
# extract and store the emails
email_details = []
for email in filtered_emails:
    # print(email)
    sender        = email.SenderName
    received_time = email.ReceivedTime
    body          = email.Body

    # convert to naive datetime by removing timezone information
    if received_time.tzinfo is not None:
        received_time = received_time.replace(tzinfo = None)
    email_details.append((sender, received_time, body))
```

```python
# dataframe the extracted emails
df = pd.DataFrame(email_details, columns = ['Sender', 'Received Time', 'Body'])

# export to Excel
pth_test = r'P:\Working Folders\Hilton\W\Reg_Tests'
df.to_excel(pth_test + '\emails_from_last_week.xlsx', index = False)

print('Exported emails to test folder')
```
