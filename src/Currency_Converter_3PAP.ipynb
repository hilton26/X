## Currency Converter

```python
# import libraries
# pip install requests
import requests # the Requests library is one of the integral parts of Python for making HTTP requests to a specified URL
```

```python
# define constants
API_KEY    = 'fca_live_Gg3B7ToiiMWPIIptaYLSGvPGaNnpcSOLVvL3gusE'
BASE_URL   = f'https://api.freecurrencyapi.com/v1/latest?apikey={API_KEY}'
CURRENCIES = ['USD', 'CAD', 'ZAR', 'GBP', 'CNY', 'EUR', 'AUD']
```

```python
# create function
# take a base currency as input and return corresponding currency rates
def convert_currency(base):
    currencies = ','.join(CURRENCIES)
    url = f'{BASE_URL}&base_currency={base}&currencies={currencies}' # multiple query parameters
    try:
        response = requests.get(url) # 'response' will contain JSON, adictionary with key-value pairs
        data = response.json()
        # print(data)
        return data        
    except:
        print('Invalid currency')
        return None    
```

```python
while True:
    curr = input('Enter the base currency (q for quit): ').upper()
    
    if curr == 'Q':
        break
    
    data = convert_currency(curr)          # format is a dictionary
    if data is None:
        continue
    else:
        del data[curr]                     # exclude the currency sought from the table of results
        for ticker, value in data.items(): # iterate over the dictionary
            print(f'{ticker}: {value}')
```
