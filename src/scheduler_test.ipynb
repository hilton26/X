```python
# from datetime import datetime
import datetime, time, schedule
```

```python
def my_script():
    """
    This is the function that will be executed.
    Replace this with the logic of your script.
    """
    print("inside my_script()")
    
    print("Running your script...", datetime.datetime.now())
```

```python
def schedule_script(start_time, end_time):
    """Schedules the script to run every two hours between given times."""
    current_time = datetime.datetime.now().time()
    if start_time <= current_time <= end_time:
        schedule.every(7).seconds.do(my_script)

    while True:
        schedule.run_pending()
        time.sleep(1)
```

```python
if __name__ == "__main__":
    start_time = datetime.time(17, 54)  # Set the start time (e.g., 9:00 AM)
    end_time   = datetime.time(20,  0)  # Set the end time (e.g., 5:00 PM)
    schedule_script(start_time, end_time)
```
