## Automated File Backup

```python
# import libraries
import os
import shutil
import datetime
import schedule
import time
```

```python
# paths
source_dir = r'C:\Users\hilton.netta\OneDrive - Prescient\Pictures'
dest_dir   = r'C:\Users\hilton.netta\OneDrive - Prescient\Backups'
```

```python
# function
def copy_folder_to_directory(source, dest):
    today = datetime.date.today()
    dest_dir = os.path.join(dest, str(today))
    
    try:
        shutil.copytree(source, dest_dir)
        print(f'Folder copied to the destination directory')
    
    except FileExistsError:
        print (f'Folder already exists in {dest}')
```

```python
# schedule    https://pypi.org/project/schedule/
schedule.every().day.at("11:55").do(lambda: copy_folder_to_directory(source_dir, dest_dir))

while True:
    schedule.run_pending()
    time.sleep(60)
```
