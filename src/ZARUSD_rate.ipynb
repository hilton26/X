```python
#
# Description:
# This script retrieves the historical exchange rate between the South African Rand (ZAR)
# and the US Dollar (USD) for a specific date provided by the user.
# It uses the 'yfinance' library to fetch data from Yahoo Finance.
#
# To run this script, you must first install the yfinance library:
# pip install yfinance
#

import yfinance as yf
from datetime import datetime, timedelta

def get_exchange_rate(target_date_str):
    """
    Fetches and returns the ZAR/USD exchange rate for a specific date.

    Args:
        target_date_str (str): The date in 'YYYY-MM-DD' format.

    Returns:
        None. It prints the exchange rate or an error message.
    """
    try:
        # Validate and parse the input date string
        target_date = datetime.strptime(target_date_str, '%Y-%m-%d').date()

        # Check if the date is in the future
        if target_date > datetime.now().date():
            print(f"Error: The date {target_date_str} is in the future. Please provide a past or current date.")
            return

        # Define the ticker symbol for ZAR to USD exchange rate
        # This represents how many USD one ZAR is worth.
        ticker_symbol = "ZARUSD=X"

        # Yahoo Finance needs a date range. To get the data for a single day,
        # we can ask for a two-day period starting from the target date.
        # This helps ensure we get a result even on days with low trading volume.
        end_date = target_date + timedelta(days=2)

        # Download the historical data for the specified ticker and date range
        print(f"\nFetching data for {ticker_symbol} on {target_date_str}...")
        data = yf.download(ticker_symbol, start=target_date, end=end_date, progress=False)

        # Check if the returned data is empty.
        # This usually happens if the market was closed (e.g., weekend, public holiday).
        if data.empty:
            print(f"No exchange rate data found for {target_date_str}.")
            print("This could be because it was a weekend or a public holiday.")
            print("Please try the nearest business day.")
            return

        # Get the closing price for the first available date in the range
        # The 'Close' price is the exchange rate at the end of the trading day.
        rate = data['Close'].iloc[0]
        
        # Display the result to the user
        print("\n--- Exchange Rate Found ---")
        print(f"Date: {target_date_str}")
        print(f"1 ZAR = {rate:.6f} USD")
        print("---------------------------")

    except ValueError:
        print("Error: Invalid date format. Please use 'YYYY-MM-DD'.")
    except Exception as e:
        print(f"An unexpected error occurred: {e}")
        print("Please check your internet connection and the ticker symbol.")

def main():
    """Main function to drive the script."""
    print("ZAR vs USD Historical Exchange Rate Finder")
    print("=========================================")
    
    # Prompt the user to enter a date
    date_input = input("Enter the date for the exchange rate (YYYY-MM-DD): ")
    
    get_exchange_rate(date_input)

if __name__ == "__main__":
    main()
```
