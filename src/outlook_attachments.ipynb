## Outlook attachments
https://stackoverflow.com/questions/39656433/how-to-download-outlook-attachment-from-python-script

```python
import datetime
import os
import win32com.client

path = r'P:\Investment Operations\GRC\Compliance\Overdrafts'
today = datetime.date.today()

outlook = win32com.client.Dispatch("Outlook.Application").GetNamespace("MAPI")
inbox = outlook.GetDefaultFolder(6) 
messages = inbox.Items

def saveattachments(subject):
    for message in messages:
        if message.Subject.startswith(subject) and message.Unread or message.Senton.date() == today:
            # body_content = message.body
            attachments = message.Attachments
            attachment = attachments.Item(1)
            for attachment in message.Attachments:
                #attachment.SaveAsFile(os.path.join(path, str(attachment)))
                attachment.SaveAsFile(os.path.join(path))
                if message.Subject == subject and message.Unread:
                    message.Unread = False
                break
```
