## __1) pypdf__

```python
# libraries, libraries!

%run utilities.ipynb # for the timediff() function
start_time = time.time()

# !pip install  pypdf
from pypdf import PdfReader
from docx import Document # https://python-docx.readthedocs.io/en/latest/
from tqdm import notebook, tqdm
import os
```

```python
# set path to the pdf to be parsed

print('Saving the pdf text to Word ...', '\n')

pim_client = r'P:\Investment Operations\Segregated Clients\Active Clients'
path_pdf   = pim_client + r'\Impala Platinum Refineries Provident Fund\Agreements\20250217_PIM_IMPREF_PMA.pdf'

# apply the pdf reader to the pdf
reader     = PdfReader(path_pdf)

lx = len(reader.pages)
print('', f'{lx} page{"" if lx == 1 else "s"} in the pdf')

# create a Word document to hold the extracted pdf text https://www.youtube.com/watch?v=re-rBqvDUPM
doc        = Document()

# https://www.w3schools.com/python/ref_string_rfind.asp#:~:text=The%20rfind()%20method%20finds,as%20the%20rindex()%20method.
doc_name   = path_pdf[path_pdf.rfind('\\') + 1:path_pdf.rfind('.')] + '.docx'

# get text from every page in the pdf and add it to the Word doc
for page in notebook.tqdm(range(len(reader.pages))):
    pg   = reader.pages[page]
    text = pg.extract_text()
    # print(text)
    doc.add_paragraph(text)
    # doc.add_page_break()

doc.save(os.path.join(r'P:\Working Folders\Hilton\W\Reg_Tests', f'{doc_name}'))

print('\n', os.path.join(r'P:\Working Folders\Hilton\W\Reg_Tests', f'{doc_name}'), '\n')

print(f'Saving the pdf text to Word completed: {timediff(start_time, time.time())}')
```

```python
# turn the pdf into an image
pth_save = r'C:\Users\hilton.netta\Downloads'
pix = page.get_pixmap()
pix.save(rf'{pth_save}\page_{page.number}.png')
```
