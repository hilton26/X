```python
import os

folder_path = r'C:\Users\hilton.netta\OneDrive - Prescient\py\gitrepo\Images'  # Replace with the actual folder path

# List all files and directories
all_entries = os.listdir(folder_path)
print(f"All entries: {all_entries}")

# List only files
files_only = [f for f in all_entries if os.path.isfile(os.path.join(folder_path, f))]
print(f"Files only: {files_only}")
```
