## Youtube Downloader

```python
# import libraries
# pip install requests
from pytube import YouTube
import tkinter as tk           # 2D graphics library taht allows one to build basic user interfaces
from tkinter import filedialog # to bring up a folder to manipulate
```

```python
# create function
url = 'https://www.youtube.com/watch?v=rJWdfDPZ9Ck'
save_path = r'C:\Users\hilton.netta\OneDrive - Prescient\Documents'
def download_video(url, save_path):
    try:
        yt = YouTube(url)
        streams = yt.streams.filter(progressive = True, file_extension = 'mp4')
        highest_res_stream = streams.get_highest_resolution()
        highest_res_stream.download(output_path = save_path)
        print('Video downloaded successfully!')
    except Exception as e:
        print(e)
```

```python
def open_file_dialogue():
    folder = filedialog.askdirectory()
    if folder:
        print(f'Selected folder: {folder}')
    return folder
```

```python
if __name__ == "__main__": # anything that happens subsequently only happens if this python file is run directly
    root = tk.Tk() # initiate a window
    root.withdraw  # hide, not deinstantiate, the window
    
    video_url = input('Please enter a Youtube url: ')
    save_dir  = open_file_dialogue()
    
    if not save_dir:
        print('Invalid save location')
    else:
        download_video(video_url, save_dir)
        
```
