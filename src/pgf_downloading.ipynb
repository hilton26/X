# Downloading PGF UCITS Hedge Share Class NAVs and Holdings

```python
print('\n\n###############################')
print('#                             #')
print('# START pgf_downloading.ipynb #')
print('#                             #')
print('###############################\n\n')
```

```python
# get input data off source sheet

import time
start_time_pgf = time.time()
start_time     = time.time()

# libraries, libraries!
import pandas as pd
import os
from pathlib import Path
from datetime import datetime
from constants import pthPy, pth_dl
from utilities import timediff, osprey, prior_working_day

print('Getting input variables ...')
# get report inputs
# get report date
df       = pd.read_excel(pthPy, sheet_name = 'arc', usecols = 'G:I')
k        = df.iloc[1,2]
rptDate  = k if isinstance(k, datetime) and not pd.isna(k) else prior_working_day(datetime.today())

# get portfolio lists as strings
clss_navs  = (',').join(df['pgf: UT prices'].dropna()) # for unit trust prices
clss_hldgs = (',').join(df['pgf: PAR-N'    ].dropna()) # for class holdings

# expected files to be downloaded
parN_nm = os.path.join(pth_dl, f'UTPS PGF_UT_prices({len(df["pgf: UT prices"].dropna())}) {rptDate.strftime("%d%b%Y")}.csv')
utP_nm  = os.path.join(pth_dl,  f'PARN PGF_Holdings({len(df["pgf: PAR-N"].dropna())}) {rptDate.strftime("%d%b%Y")}.csv')
print(f'{rptDate.strftime("%a %d %b %Y")} PGF files to be downloaded:\n  {parN_nm}\n  {utP_nm}\n')

# check inputs
print(f' Report date:          {rptDate.date().strftime("%a %d %b %Y")}', '\n', f'Class NAV codes:      {clss_navs}', \
      '\n', f'Class holdings codes: {clss_hldgs}')
print(f'\n{timediff(start_time, time.time())} getting input variables','\n')
```

```python
# get class prices / NAVs in csv format
start_time = time.time()
print('Getting the unit trust prices report ...')

try:
    start_time_prices = time.time()
    print("... downloading PGF unit trust prices")
    # check if the file was already downloaded before running osprey()
    if os.path.isfile(parN_nm):
        print(f"  {parN_nm} already exists")
        pass
    else:
        osprey(
            "utps", clss_navs, rptDate, rptDate, "PGF_UT_prices", "csv")
        print(
            f"  {timediff(start_time_prices, time.time())} getting PGF unit trust prices"
        )
except Exception as e:
    print(e)

print(f'{timediff(start_time, time.time())} getting the unit trust prices report','\n')
```

```python
# get class holdings in csv format
start_time = time.time()
print('Getting PGF class holdings ...')

try:
    start_time_navs = time.time()
    print("... downloading PGF holdings")
    # check if the file was already downloaded before running osprey()
    if os.path.isfile(utP_nm):
        print(f"  {utP_nm} already exists")
        pass
    else:
        osprey("parn", clss_hldgs, rptDate, rptDate, "PGF_Holdings", "csv")
        print(f"  {timediff(start_time_navs, time.time())} getting PGF holdings")
except Exception as e:
    print(e)

print(f'{timediff(start_time  , time.time())} getting the PGF class holdings\n')
print(f'{timediff(start_time_pgf, time.time())} roundtrip time to get unit trust prices and holdings reports\n')
```

```python
# report which files, if any, were downloaded
print(f"{rptDate.strftime('%a %d %B %Y')} downloads")
print(f"  {parN_nm} which {'exists' if os.path.exists(parN_nm) else 'does not exist'}")
print(f"  {utP_nm} which {'exists' if os.path.exists(utP_nm) else 'does not exist'}")
```
