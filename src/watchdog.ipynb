```python
'''Gemini 12 Aug 2025 

Query: "python dataframe a file as soon as it is in 
the downloads folder", 

Response: To automatically load a newly downloaded file in the "Downloads" folder into
a Pandas DataFrame, you can use the watchdog library to monitor the folder 
for changes and then use pandas.read_csv (or a similar function for 
other file types) to load the file.'''

import os
import pandas as pd
from watchdog.observers import Observer
from watchdog.events import FileSystemEventHandler
import time

class MyHandler(FileSystemEventHandler):
    def on_created(self, event):
        if not event.is_directory:
            file_path = event.src_path
            print(f"New file detected: {file_path}")

            # Determine the file type and load accordingly
            if file_path.endswith('.csv'):
                try:
                    df = pd.read_csv(file_path)
                    print("CSV file loaded into DataFrame:")
                    print(df.head())
                except Exception as e:
                    print(f"Error loading CSV: {e}")
            elif file_path.endswith(('.xlsx', '.xls')):
                try:
                    df = pd.read_excel(file_path)
                    print("Excel file loaded into DataFrame:")
                    print(df.head())
                except Exception as e:
                    print(f"Error loading Excel: {e}")
            else:
                print("Unsupported file type for DataFrame loading.")

if __name__ == "__main__":
    # Get the user's Downloads folder path
    downloads_folder = os.path.join(os.path.expanduser('~'), 'Downloads')
    print(f"Monitoring: {downloads_folder}")

    event_handler = MyHandler()
    observer = Observer()
    observer.schedule(event_handler, downloads_folder, recursive=False)
    observer.start()

    try:
        while True:
            time.sleep(1)
    except KeyboardInterrupt:
        observer.stop()
    observer.join()
```
