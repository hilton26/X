```python
import schedule
import time
from datetime import datetime
```

```python
def my_job():
    print(f"Job ran at {datetime.now()}")
```

```python
import schedule, time

desired_weekday = "tuesday" # This can be dynamically set

# Schedule the job
if hasattr(schedule.every(), desired_weekday):
    getattr(schedule.every(), desired_weekday).at("10:00").do(my_job)
else:
    print(f"Invalid weekday: {desired_weekday}")
```
