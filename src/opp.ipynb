Gemini query: "how do i execute python script by pushing a button on an htmal page"

```python
#!pip install Flask


from flask import Flask, jsonify

# Create a Flask app instance
app = Flask(__name__)

# This is the function your button will trigger
def do_something():
    """
    This is your python script.
    It can do anything, like calculations, file operations,
    or database lookups.
    """
    print("The Python script is running!")
    # For this example, we'll just return a simple message.
    return "Python function executed successfully!"

@app.route('/run_script')
def run_script():
    """
    This function is linked to the '/run_script' URL.
    When the browser requests this URL, Flask runs this function.
    """
    result = do_something()
    # Return the result as a JSON object
    return jsonify(message=result)

# This makes the script runnable
if __name__ == '__main__':
    # Runs the Flask app on http://127.0.0.1:5000/
    app.run(debug=True)
```
