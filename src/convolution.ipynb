```python
# !pip install scipy
import numpy as np
import timeit
import scipy

arr1 = np.random.random(100000) # 100k random values
arr2 = np.random.random(100000) # 100k random values
```

```python
%%timeit
np.convolve(arr1, arr2)
```

```python
%%timeit
scipy.signal.fftconvolve(arr1, arr2)
```
