### Optimisation

#### Minimize $f(x) = (x-3)^2$

```python
def f(x):
    return (x-3)**2

res = minimize(f, 2) # 2 is the initial guess
res
```

```python
res.x
```

Minimise $f(x, y) = (x-1)^2 + (y-2.5)^2$ subject to
$$x-2y+2>=0$$
$$-x-2y+6>=0$$
$$-x+2y+2>=0$$
$$x>=0$$
$$y>=0$$

* 2D function takes in a vector $x[.]$
* Constraints i are specified as $g_i(x)>=0$ as a tuple of dictionaries
* Bounds are specified as rectangular

```python
f = lambda x: (x[0] - 1)**2 + (x[1] - 2.5)**2
# e.g., evaluate f at x = 3 (or x[0] = 3) and y = 7 (or x[1] = 7):
f([3, 7])
```

```python
# set the constraints and boundaries
cons = ({'type': 'ineq', 'fun': lambda x:  x[0]-2*x[1]+2},
        {'type': 'ineq', 'fun': lambda x: -x[0]-2*x[1]+6},
        {'type': 'ineq', 'fun': lambda x: -x[0]+2*x[1]+2})
bnds = ((0, None), (0, None)) # 0, is the lower bound and 'None' means upside is unbounded
```

```python
# apply the minimise function
res = minimize(f, (2,0), bounds = bnds, constraints = cons)
res.x
```
