```python
# date discerner 2

import re
from datetime import datetime

YEAR = r"(?:19|20|21)\d{2}"
YYYYMMDD = re.compile(rf"^(?P<year>{YEAR})(?P<month>\d{{2}})(?P<day>\d{{2}})$")
DDMMYYYY = re.compile(rf"^(?P<day>\d{{2}})(?P<month>\d{{2}})(?P<year>{YEAR})$")

def parse_8digit_date(s: str) -> datetime.date:
    for pattern in (YYYYMMDD, DDMMYYYY):
        m = pattern.match(s)
        if not m:
            continue
        try:
            return datetime(int(m.group("year")), int(m.group("month")), int(m.group("day"))).date()
        except ValueError:
            continue
    raise ValueError(f"'{s}' doesn't match a valid YYYYMMDD or DDMMYYYY date")


if __name__ == "__main__":
    for test in ["20260610", "10062026", "19991231", "31121999"]:
        print(test, "->", parse_8digit_date(test))
```

```python
# date discerner 1

import re
from datetime import datetime

YEAR = r"(?:19|20|21)\d{2}"

PATTERN = re.compile(
    rf"^(?:(?P<year1>{YEAR})(?P<day1>\d{{2}})(?P<month1>\d{{2}})"
    rf"|(?P<day2>\d{{2}})(?P<month2>\d{{2}})(?P<year2>{YEAR}))$"
)

def parse_8digit_date(s: str) -> datetime.date:
    m = PATTERN.match(s)
    if not m:
        raise ValueError(f"'{s}' doesn't match an 8-digit DDMMYYYY/YYYYDDMM pattern")

    if m.group("year1"):
        year, day, month = m.group("year1"), m.group("day1"), m.group("month1")
    else:
        year, day, month = m.group("year2"), m.group("day2"), m.group("month2")

    return datetime(int(year), int(month), int(day)).date()


if __name__ == "__main__":
    for test in ["20260610", "10062026", "31121999", "19991231"]:
        print(test, "->", parse_8digit_date(test))

```
