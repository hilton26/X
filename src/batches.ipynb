# ChatGPT prompt 2 Dec 2025: "python script to split a list of items into batches of 10"

```python
def batch_list(items, batch_size=10):
    return [items[i:i + batch_size] for i in range(0, len(items), batch_size)]

# Example usage:
my_list = list(range(1, 7))  # sample list
batches = batch_list(my_list)

for b in batches:
    print(b)
```
