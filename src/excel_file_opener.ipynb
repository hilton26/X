```python
# create a time difference function
import time
from datetime import datetime

def timediff(start, end, decimals = 1):
    if   int((end - start) / 3600) > 0: # non-zero hours
        return str(int(  (end - start) / 3600))           + 'hr '  + \
               str(int(  (end - start) /   60))           + 'min ' + \
               str(round((end - start) %   60, decimals)) + 'sec'
    elif int((end - start) /   60) > 0: # non-zero hours and minutes
        return str(int(  (end - start) /   60))           + 'min ' + \
               str(round((end - start) %   60, decimals)) + 'sec'
    else:
        return str(round((end - start) %   60, decimals)) + 'sec'
```

```python
# utility functions
start_time = time.time()
print(f'Making utility functions ...')

# utility function to open an excel file, .xls or .xlsx
def open_xl_file(file_name_and_path):
    import win32com.client as win32 # library to convert xls to xlsx
    excel               = win32.gencache.EnsureDispatch('Excel.Application')
    excel.DisplayAlerts = False # suppress the warning dialogue
    excel.Workbooks.Open(file_name_and_path)
    excel.DisplayAlerts = True # unsuppress the warning dialogue
    
print(f'Making utility functions completed: {timediff(start_time, time.time())}', '\n')
```

```python
# get report variables
start_time = time.time()
print(f'Getting report variables ...')

#get report date and fund names from the py_report.xlsm sheet
import xlwings as xw
wb_py    = xw.Book(r'P:\Investment Operations\GRC\Compliance\Daily\py_reports.xlsm') # open calc workbook as an object
rptDate  = wb_py.sheets['dervs'].range('B1').value.strftime('%d%b%Y')  
funds_py = [f for f in wb_py.sheets['dervs'].range('A:A')[1:].value if f is not None]
wb_py.close()

# get list of completed derivative calc files in the folder P:\Investment Operations\GRC\Compliance\Derivative Cover
dv_files = [f for f in os.listdir(pth_derv) if f[-24:] == f'Derv Calc {rptDate}.xlsx']
funds    = [file[:file.find(' ')] for file in dv_files]
lst_dv   = ", ".join([file[:file.find(' ')] for file in dv_files])
print(f' {len(dv_files)} derv calc files for {rptDate} in the derv reporting folder over {len(funds_py)} funds')

print(f'Getting report variables completed: {timediff(start_time, time.time())}', '\n')
```

```python
# open the summary derivative calc sheet
start_time = time.time()
print(f'Opening the summary derivative calc file ...')

wb_main = xw.Book(fr'P:\Investment Operations\GRC\Compliance\Derivative Cover\Derv {rptDate}.xlsx')
print(f' Derv {rptDate}.xlsx')

print(f'Opening the summary derivative calc file completed: {timediff(start_time, time.time())}', '\n')
```

```python
# the loop
start_time = time.time()
print(f'Opening and then closing each derivative calc file')

# open the summary derv calc workbook as an object
for fund in tqdm(funds):
    wb     = xw.Book(fr'P:\Investment Operations\GRC\Compliance\Derivative Cover\{fund} Derv Calc {rptDate}.xlsx')
    #time.sleep(1) # sleep for 1 second before closing the file
    wb.close()

print(f'Opening and then closing each derivative calc file ...')
```
