```python
# libraries, libraries!

from PIL import Image
import pytesseract
import fitz  # PyMuPDF
import re
import os
from tqdm import notebook, tqdm  # notebook version of tqdm

# set the path to the Tesseract executable if needed
pytesseract.pytesseract.tesseract_cmd = r'C:\Users\hilton.netta\AppData\Local\Programs\Tesseract-OCR\tesseract.exe'
```

```python
# create functions

def extract_text_from_image(image_path):
    """
    Extracts text from an image file using OCR.
    """
    try:
        with Image.open(image_path) as img:
            text = pytesseract.image_to_string(img)
            return text
    except FileNotFoundError:
        return f"Error: Image file not found at {image_path}"
    except Exception as e:
        return f"An error occurred with the image: {e}"

def extract_text_from_pdf(pdf_path):
    """
    Extracts text from a PDF file.
    """
    text = ""
    try:
        with fitz.open(pdf_path) as doc:
            for page in doc:
                text += page.get_text()
        return text
    except FileNotFoundError:
        return f"Error: PDF file not found at {pdf_path}"
    except Exception as e:
        return f"An error occurred with the PDF: {e}"

def extract_from_file(file_path):
    """
    Extracts text and numbers from a file based on its extension.
    """
    if not os.path.exists(file_path):
        return f"Error: File not found at {file_path}", []

    file_extension = os.path.splitext(file_path)[1].lower()
    
    # Process Images
    if file_extension in ['.jpg', '.jpeg', '.png', '.tiff', '.bmp']:
        raw_text = extract_text_from_image(file_path)
    
    # Process PDFs
    elif file_extension == '.pdf':
        raw_text = extract_text_from_pdf(file_path)
    
    # Unsupported file type
    else:
        return "Error: Unsupported file type.", []

    # Find numbers using regular expressions
    numbers = re.findall(r'\d+', raw_text)
    
    return raw_text, numbers

# if __name__ == '__main__':
#     # # Example usage for an image file
#     # image_file = 'example_image.jpg'
#     # text_from_image, numbers_from_image = extract_from_file(image_file)
#     # print(f"--- Processing Image: {image_file} ---")
#     # print("Extracted Text:\n", text_from_image)
#     # print("Extracted Numbers:\n", numbers_from_image)
#     # print("-" * 30)

#     # # Example usage for a PDF file
#     # pdf_file = 'example_document.pdf'
#     # text_from_pdf, numbers_from_pdf = extract_from_file(pdf_file)
#     # print(f"--- Processing PDF: {pdf_file} ---")
#     # print("Extracted Text:\n", text_from_pdf)
#     # print("Extracted Numbers:\n", numbers_from_pdf)
#     # print("-" * 30)
```
